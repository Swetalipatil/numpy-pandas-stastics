#### Q1. Arithmetic Mean:- The weekly sales (in units) of a retail store were: 48, 52, 61, 57, 49, 63, 70
> **Calculate**:
> 1. The arithmetic mean.
> 2. The deviation of each observation from the mean.
> 3. Verify that the sum of deviations from the mean is zero.


```python
sales = [48, 52, 61, 57, 49, 63, 70]
total = 48 + 52 + 61 + 57 + 49 + 63 + 70
n = 7
mean = total / n

print("Deviation of Each Observation")
d1 = sales[0] - mean
print("48 -", round(mean, 2), "=", round(d1, 2))

d2 = sales[1] - mean
print("52 -", round(mean, 2), "=", round(d2, 2))

d3 = sales[2] - mean
print("61 -", round(mean, 2), "=", round(d3, 2))

d4 = sales[3] - mean
print("57 -", round(mean, 2), "=", round(d4, 2))

d5 = sales[4] - mean
print("49 -", round(mean, 2), "=", round(d5, 2))

d6 = sales[5] - mean
print("63 -", round(mean, 2), "=", round(d6, 2))

d7 = sales[6] - mean
print("70 -", round(mean, 2), "=", round(d7, 2))

sum_deviation = d1 + d2 + d3 + d4 + d5 + d6 + d7
print("Sum of Deviations =", round(sum_deviation, 2))

# verify
if round(sum_deviation, 2) == 0:
    print("Verification: The sum of deviations from the mean is ZERO.")
else:
    print("Verification: The sum of deviations is approximately ZERO because the mean is rounded.")
```

#### Q21. Coefficient of Variation:-

> Two investment portfolios have the following statistics:
> - Portfolio A: Mean return = 12%, Standard deviation = 2%
> - Portfolio B: Mean return = 18%, Standard deviation = 4%

> Calculate the coefficient of variation for each portfolio and determine which investment is more
consistent.

```python
mean_A = 12
std_dev_A = 2
mean_B = 18
std_dev_B = 4

cv_A = (std_dev_A / mean_A) * 100

cv_B = (std_dev_B / mean_B) * 100

print(f"Coefficient of Variation for Portfolio A = {cv_A:.2f}%")
print(f"Coefficient of Variation for Portfolio B = {cv_B:.2f}%")

if cv_A < cv_B:
    print("Portfolio A is more consistent.")
elif cv_B < cv_A:
    print("Portfolio B is more consistent.")
else:
    print("Both portfolios have the same consistency.")
```

#### Q23. Effect of Multiplying by a Constant:- The observations are: 5, 8, 11, 14, 17. Each observation is multiplied by 4. Determine the new:
> 1. Mean
> 2. Variance
> 3. Standard deviation

```python
observations = [5, 8, 11, 14, 17]
constant_multiplier = 4
original_mean = statistics.mean(observations)
original_variance = statistics.variance(observations)
original_std_dev = statistics.stdev(observations)
new_mean = original_mean * constant_multiplier
new_variance = original_variance * (constant_multiplier ** 2)
new_std_dev = original_std_dev * abs(constant_multiplier)

print(f"New Mean (after multiplying by {constant_multiplier}) = {new_mean}")
print(f"New Variance (after multiplying by {constant_multiplier}) = {new_variance}")
print(f"New Standard Deviation (after multiplying by {constant_multiplier}) = {new_std_dev:.2f}")
```

### Bayes' Theorem:-

#### Q47. A disease affects 2% of a population. A diagnostic test has:
> - Sensitivity = 95%
> - Specificity = 98%

> A person tests positive. Find the probability that the person actually has the disease.

```python
disease = 0.02
no_disease = 0.98

sensitivity = 0.95
specificity = 0.98
false_positive = 1 - specificity

# Bayes' Theorem
positive = (sensitivity * disease) + (false_positive * no_disease)
probability = (sensitivity * disease) / positive
print("Probability that the person actually has the disease =", probability)
print("Percentage =", probability * 100)
```

#### Q48. A spam filter correctly identifies:
> - 96% of spam emails.
> - 93% of legitimate emails.

> Suppose 25% of all incoming emails are spam.
> If an email is classified as spam, what is the probability that it is actually spam?

```python
spam = 0.25
legitimate = 0.75

spam_correct = 0.96
legitimate_correct = 0.93
false_positive = 1 - legitimate_correct
total_spam = (spam_correct * spam) + (false_positive * legitimate)

# Bayes' Theorem
answer = (spam_correct * spam) / total_spam
print("Probability that the email is actually spam =", answer)
print("Percentage =", answer * 100, "%")
```

#### Q100. A machine is designed to fill bottles with 1 litre of liquid.
> A sample of 100 bottles has:
> - Mean = 0.985 litre
> - Population standard deviation = 0.04 litre

> At the 5% significance level:
> 1. State the null and alternative hypotheses.
> 2. Compute the Z statistic.
> 3. Determine the p-value.
> 4. State whether the null hypothesis should be rejected.
> 5. Construct a 95% confidence interval for the population mean.
> 6. Interpret the confidence interval in the context of the problem.

```python
x_bar = 0.985
mu = 1
sigma = 0.04
n = 100

z = (x_bar - mu) / (sigma / math.sqrt(n))
print("Z =", round(z, 2))

if abs(z) > 1.96:
    print("Reject H0")
else:
    print("Fail to Reject H0")

margin = 1.96 * (sigma / math.sqrt(n))
lower = x_bar - margin
upper = x_bar + margin

print("95% CI = (", round(lower,4), ",", round(upper,4), ")")
```
