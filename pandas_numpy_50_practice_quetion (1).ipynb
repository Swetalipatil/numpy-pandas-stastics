### NP40. Apply rolling computation using NumPy.
> 💡 **Hint:** Use a list comprehension: `[np.mean(arr[max(0,i-w+1):i+1]) for i in range(len(arr))]` to manually compute a rolling mean with window `w`. Compare with `df["col"].rolling(w).mean()`.


```python
df = pd.DataFrame({
    "col": [10, 20, 30, 40, 50, 60, 70]
})

print("Original DataFrame:")
print(df)

arr = df["col"].values

w = 3

df["NumPy_Rolling_Mean"] = [
    np.mean(arr[max(0, i - w + 1):i + 1])
    for i in range(len(arr))
]

df["Pandas_Rolling_Mean"] = df["col"].rolling(
    w,
    min_periods=1
).mean()

print("DataFrame After Rolling Computation:")
print(df)
```

### NP42. Detect and replace outliers using NumPy.
> 💡 **Hint:** Compute Q1, Q3 with `np.percentile`. Define bounds as `Q1 - 1.5*IQR` and `Q3 + 1.5*IQR`. Use `np.clip(arr, lo, hi)` to replace outliers with boundary values.


```python
df = pd.DataFrame({
    "Marks": [10, 20, 30, 40, 50, 60, 70, 200]
})
print("Original DataFrame:")
print(df)
arr = df["Marks"].values
Q1 = np.percentile(arr, 25)
Q3 = np.percentile(arr, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(":", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

clean_arr = np.clip(
    arr,
    lower_bound,
    upper_bound
)

df["Clean_Marks"] = clean_arr

print("DataFrame After Replacing Outliers:")
print(df)
```

```python

df = pd.DataFrame({
    "Marks": [10, 20, 30, 40, 50]
})

print("Original DataFrame:")
print(df)

arr = df["Marks"].values


min_value = np.min(arr)
max_value = np.max(arr)

df["MinMax_Scaled"] = (
    (arr - min_value) /
    (max_value - min_value)
)


mean_value = np.mean(arr)
std_value = np.std(arr)

df["ZScore_Scaled"] = (
    (arr - mean_value) /
    std_value
)

print("DataFrame After Scaling:")
print(df)
```

## NP44. Engineer new features using NumPy.
> 💡 **Hint:** Create polynomial features (`np.square`, `np.power`), log features (`np.log1p`), interaction features (column1 × column2), and binned features (`np.digitize`).


```python

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000]
})

print("Original DataFrame:")
print(df)
df["Age_Squared"] = np.square(df["Age"])
df["Age_Cubed"] = np.power(df["Age"], 3)
df["Log_Salary"] = np.log1p(df["Salary"])
df["Age_Salary_Interaction"] = (
    df["Age"] * df["Salary"]
)
bins = [25, 35]
df["Age_Bin"] = np.digitize(
    df["Age"],
    bins
)
print("DataFrame After Feature Engineering:")
print(df)
```


### NP49. Perform a mini data analysis project.
> 💡 **Hint:** Follow these steps: (1) Generate data with NumPy, (2) Store in DataFrame, (3) Clean with Pandas, (4) Engineer features with NumPy, (5) GroupBy analysis, (6) Detect outliers, (7) Export summary.

```python
np.random.seed(42)
n = 50
df = pd.DataFrame({
    "Student_ID": np.arange(1, n + 1),
    "Department": np.random.choice(["DS", "ML", "AI"], n),
    "Marks": np.random.randint(40, 101, n),
    "Attendance": np.random.randint(60, 101, n)
})
print("Original Dataset:")
print(df.head())


df.loc[5, "Marks"] = np.nan
df.loc[15, "Marks"] = np.nan
df.loc[25, "Marks"] = np.nan

print("Missing Values:")
print(df.isnull().sum())


mean_marks = df["Marks"].mean()

df["Marks"] = df["Marks"].fillna(mean_marks)

print("Dataset After Cleaning:")
print(df.head())

df["Status"] = np.where(
    df["Marks"] >= 75,
    "Pass",
    "Fail"
)

conditions = [
    df["Marks"] >= 90,
    (df["Marks"] >= 75) & (df["Marks"] < 90),
    (df["Marks"] >= 60) & (df["Marks"] < 75)
]
grades = ["A", "B", "C"]

df["Grade"] = np.select(
    conditions,
    grades,
    default="D"
)
df["Attendance_Category"] = np.where(
    df["Attendance"] >= 85,
    "Good",
    "Needs Improvement"
)

print("Dataset After Feature Engineering:")
print(df.head())


print("Average Marks by Department:")
print(
    df.groupby("Department")["Marks"].mean()
)

print("Department Summary:")
department_summary = df.groupby("Department").agg({
    "Marks": ["mean", "max", "min"],
    "Attendance": ["mean", "max", "min"]
})

print(department_summary)

Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Marks"] < lower_bound) |
    (df["Marks"] > upper_bound)
]

print("Outliers:")
print(outliers)

print("Grade Distribution:")
print(df["Grade"].value_counts())

print("Pass/Fail Count:")
print(df["Status"].value_counts())

print("Top 5 Students by Marks:")
top_students = df.sort_values(
    by="Marks",
    ascending=False
).head(5)

print(top_students)

df.to_csv(
    "cleaned_student_data.csv",
    index=False
)

department_summary.to_csv(
    "department_summary.csv"
)
print("Files Saved Successfully!")
print("1. cleaned_student_data.csv")
print("2. department_summary.csv")
```

### NP50. Build an end-to-end pipeline using NumPy + Pandas.
> 💡 **Hint:** Chain all steps together:  
> 1. `np.random.seed(X)` → generate raw data  
> 2. `np.clip` → clean invalid values  
> 3. `np.log`, `np.square`, `np.where` → feature engineering  
> 4. `df.groupby().agg(...)` → statistical analysis  
> 5. IQR + `np.clip` → outlier removal  
> 6. `df[features].to_numpy(dtype=np.float32)` → ML feature matrix  
> 7. Export with `df.to_csv(...

```python
# NP50 - End-to-End Pipeline Using NumPy + Pandas
import numpy as np
import pandas as pd

# Step 2 : Generate random data
np.random.seed(42)

n = 20

# Create DataFrame
df = pd.DataFrame({
    "Age": np.random.randint(18, 60, n),
    "Salary": np.random.randint(10000, 100000, n),
    "Department": np.random.choice(["HR", "IT", "Sales"], n)
})

# Display original dataset
print(" ORIGINAL DATA")
print(df)

# Step 3 : Data Cleaning
# Salary should be between 15000 and 90000


df["Salary"] = np.clip(df["Salary"], 15000, 90000)

print("AFTER CLEANING")
print(df)

# Step 4 : Feature Engineering
# Create new useful columns


# Log of Salary
df["Log_Salary"] = np.log(df["Salary"])

# Square of Age
df["Age_Squared"] = np.square(df["Age"])

# High Salary
# If salary > 50000 then 1 otherwise 0
df["High_Salary"] = np.where(df["Salary"] > 50000, 1, 0)

print("AFTER FEATURE ENGINEERING ")
print(df)
# Step 5 : Group By Analysis


summary = df.groupby("Department").agg({
    "Salary": ["mean", "min", "max"],
    "Age": ["mean", "count"]
})

print(" GROUP BY ANALYSIS ")
print(summary)

# Step 6 : Outlier Detection using IQR
# Convert Salary column into NumPy array
salary = df["Salary"].values

# Find Quartiles
Q1 = np.percentile(salary, 25)
Q3 = np.percentile(salary, 75)

# Find IQR
IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

print("\nQ1 :", Q1)
print("Q3 :", Q3)
print("IQR :", IQR)
print("Lower Limit :", lower_limit)
print("Upper Limit :", upper_limit)
# Step 7 : Remove Outliers
df["Salary_Clean"] = np.clip(
    salary,
    lower_limit,
    upper_limit
)

print(" AFTER OUTLIER TREATMENT ")
print(df[["Salary", "Salary_Clean"]])

# Step 8 : Select Features for Machine Learning
features = [
    "Age",
    "Salary_Clean",
    "Log_Salary",
    "Age_Squared"
]
# Convert DataFrame into NumPy array
X = df[features].to_numpy(dtype=np.float32)
print("MACHINE LEARNING FEATURE MATRIX ")
print(X)
print("Shape of Feature Matrix")
print(X.shape)

# Step 9 : Export Dataset
df.to_csv("processed_dataset.csv", index=False)
print("Dataset saved successfully!")
# Step 10 : Final Dataset

print( "FINAL DATASET")
print(df)
```
