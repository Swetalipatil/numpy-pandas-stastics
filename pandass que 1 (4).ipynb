```python
# 1
import pandas as pd
numbers = [10, 20, 30, 40, 50]
s = pd.Series(numbers)
print(s)
```

```python
# 2
marks = {
    "Rahul": 80,
    "Priya": 90,
    "Amit": 75
}

s = pd.Series(marks)

print(s)
```

```python
# 3
students = [
    {"Name": "Rahul", "Age": 20},
    {"Name": "Priya", "Age": 21},
    {"Name": "Amit", "Age": 22}
]

df = pd.DataFrame(students)

print(df)
```

```python
# 4
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"]
})

df["Country"] = "India"

print(df)
```

```python
# 5
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"]
})

ages = pd.Series([20, 21, 22])

df["Age"] = ages

print(df)
```

```python
# 6 one coln
df = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Age": [20, 21]
})

df = df.rename(columns={"Age": "Student_Age"})

print(df)




# rename multiple coln

df = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Age": [20, 21]
})

df = df.rename(columns={
    "Name": "Student_Name",
    "Age": "Student_Age"
})

print(df)
```

```python
# 7
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 22]
})

print(df)

# select single row
print(df.loc[0])
# select multiple row
print(df.loc[[0, 2]])
# select conditional row
print(df.loc[df["Age"] > 20])
```

```python
# 8

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 22]
})

print("Shape:")
print(df.shape)

print("Data Types:")
print(df.dtypes)
```

```python
# 10
import numpy as np

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Marks": [80, np.nan, 75]
})

print("Before fillna():")
print(df)

df["Marks"] = df["Marks"].fillna(0)

print("\nAfter fillna():")
print(df)
```
